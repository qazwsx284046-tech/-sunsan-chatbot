```python
from google.colab import auth
from google.cloud import bigquery
from google.api_core.exceptions import Conflict

# 1. 구글 계정 인증 (팝업 창에서 허용)
auth.authenticate_user()


# 2. 프로젝트 및 데이터셋 설정
PROJECT_ID = "project-bb3dade7-528c-44c4-936"
DATASET_ID = "youtube_analysis" # 생성할 데이터셋 이름

client = bigquery.Client(project=PROJECT_ID)

# 3. 데이터셋 생성 (이미 존재하면 건너뜀)
dataset_ref = f"{PROJECT_ID}.{DATASET_ID}"
dataset = bigquery.Dataset(dataset_ref)
dataset.location = "asia-northeast3" # 서울 리전 설정

try:
    client.create_dataset(dataset, timeout=30)
    print(f"✅ 데이터셋 생성 완료: {dataset_ref}")
except Conflict:
    print(f"ℹ️ 데이터셋이 이미 존재합니다: {dataset_ref}")

# 4. 테이블 생성 DDL 쿼리 작성 (파티셔닝 & 클러스터링 완벽 적용)

# [테이블 1] videos_metadata
create_videos_table_sql = f"""
CREATE TABLE IF NOT EXISTS `{PROJECT_ID}.{DATASET_ID}.videos_metadata` (
  video_id STRING,
  channel_id STRING,
  title STRING,
  published_at TIMESTAMP,
  view_count INT64,
  like_count INT64,
  comment_count INT64,
  tags ARRAY<STRING>,
  collected_at TIMESTAMP
)
PARTITION BY DATE(published_at)
CLUSTER BY channel_id;
"""

# [테이블 2] comments_sentiment
create_comments_table_sql = f"""
CREATE TABLE IF NOT EXISTS `{PROJECT_ID}.{DATASET_ID}.comments_sentiment` (
  comment_id STRING,
  video_id STRING,
  text_original STRING,
  text_cleaned STRING,
  author_name STRING,
  published_at TIMESTAMP,
  like_count INT64,
  is_reply BOOLEAN,
  sentiment_label STRING,
  sentiment_score FLOAT64,
  sentiment_magnitude FLOAT64,
  analyzed_at TIMESTAMP
)
PARTITION BY DATE(published_at)
CLUSTER BY video_id, sentiment_label;
"""

# 5. 테이블 생성 쿼리 실행
print("\n⏳ 테이블 생성을 시작합니다...")

client.query(create_videos_table_sql).result()
print("✅ videos_metadata 테이블 생성 완료 (기준: published_at 파티셔닝)")

client.query(create_comments_table_sql).result()
print("✅ comments_sentiment 테이블 생성 완료 (기준: published_at 파티셔닝, video_id/sentiment_label 클러스터링)")
```

```python
# 1. 필요 라이브러리 설치 및 Import
!pip install --upgrade google-api-python-client google-cloud-bigquery

from datetime import datetime, timezone
from google.colab import auth
from google.cloud import bigquery
from googleapiclient.discovery import build

# 구글 계정 인증 (Colab)
auth.authenticate_user()

# 2. 기본 설정
PROJECT_ID = "project-bb3dade7-528c-44c4-936"
DATASET_ID = "youtube_analysis"
YOUTUBE_API_KEY = "AIzaSyDm5kn8ggGuGh6LAIr2BH80K5hwKA5DDN"

SEARCH_KEYWORD = "오디세이"
MAX_VIDEOS = 10  # 수집할 상위 영상 수
MAX_COMMENTS_PER_VIDEO = 100  # 영상당 수집할 최대 댓글 수

# 클라이언트 생성
bq_client = bigquery.Client(project=PROJECT_ID)
youtube = build("youtube", "v3", developerKey=YOUTUBE_API_KEY)

# 3. 데이터 수집 함수 정의
def get_top_videos(keyword, max_results=10):
    """키워드로 검색하여 상위 영상 정보 수집"""
    search_response = (
        youtube.search()
        .list(q=keyword, part="id,snippet", type="video", order="relevance", maxResults=max_results)
        .execute()
    )

    video_ids = [item["id"]["videoId"] for item in search_response.get("items", [])]
    if not video_ids:
        return []

    # 영상 상세 정보 및 통계 조회
    videos_response = (
        youtube.videos()
        .list(id=",".join(video_ids), part="snippet,statistics")
        .execute()
    )

    collected_at = datetime.now(timezone.utc).isoformat()
    videos_data = []

    for item in videos_response.get("items", []):
        snippet = item.get("snippet", {})
        stats = item.get("statistics", {})

        videos_data.append(
            {
                "video_id": item["id"],
                "channel_id": snippet.get("channelId"),
                "title": snippet.get("title"),
                "published_at": snippet.get("publishedAt"),
                "view_count": int(stats.get("viewCount", 0)),
                "like_count": int(stats.get("likeCount", 0)),
                "comment_count": int(stats.get("commentCount", 0)),
                "tags": snippet.get("tags", []),
                "collected_at": collected_at,
            }
        )

    return videos_data


def get_video_comments(video_id, max_results=100):
    """특정 영상의 댓글 목록 수집"""
    comments_data = []
    next_page_token = None

    while len(comments_data) < max_results:
        try:
            request = youtube.commentThreads().list(
                part="snippet",
                videoId=video_id,
                textFormat="plainText",
                maxResults=min(100, max_results - len(comments_data)),
                pageToken=next_page_token,
            )
            response = request.execute()

            for item in response.get("items", []):
                snippet = item["snippet"]["topLevelComment"]["snippet"]
                comments_data.append(
                    {
                        "comment_id": item["id"],
                        "video_id": video_id,
                        "text_original": snippet.get("textOriginal"),
                        "text_cleaned": None,  # 전처리 단계에서 채울 필드
                        "author_name": snippet.get("authorDisplayName"),
                        "published_at": snippet.get("publishedAt"),
                        "like_count": int(snippet.get("likeCount", 0)),
                        "is_reply": False,
                        "sentiment_label": None,  # 감성 분석 모델 적용 단계에서 채울 필드
                        "sentiment_score": None,
                        "sentiment_magnitude": None,
                        "analyzed_at": None,
                    }
                )

            next_page_token = response.get("nextPageToken")
            if not next_page_token:
                break

        except Exception as e:
            # 댓글이 비활성화되어 있거나 권한 오류 발생 시 건너뜀
            print(f"⚠️ 영상 ({video_id}) 댓글 수집 중 건너뜀: {e}")
            break

    return comments_data


# 4. 메인 수집 및 BigQuery 적재 프로세스
print(f"🔍 '{SEARCH_KEYWORD}' 관련 상위 {MAX_VIDEOS}개 영상 수집 시작...")
videos = get_top_videos(SEARCH_KEYWORD, max_results=MAX_VIDEOS)

if not videos:
    print("❌ 수집된 영상이 없습니다.")
else:
    # 4-1. videos_metadata 테이블 적재
    videos_table_id = f"{PROJECT_ID}.{DATASET_ID}.videos_metadata"
    errors = bq_client.insert_rows_json(videos_table_id, videos)

    if errors == []:
        print(f"✅ 'videos_metadata' 테이블에 {len(videos)}개 영상 정보 적재 완료")
    else:
        print(f"❌ 영상 적재 중 오류 발생: {errors}")

    # 4-2. comments_sentiment 테이블 적재
    all_comments = []
    for idx, v in enumerate(videos, 1):
        v_id = v["video_id"]
        print(f"💬 [{idx}/{len(videos)}] 영상 ({v_id}) 댓글 수집 중...")
        comments = get_video_comments(v_id, max_results=MAX_COMMENTS_PER_VIDEO)
        all_comments.extend(comments)

    if all_comments:
        comments_table_id = f"{PROJECT_ID}.{DATASET_ID}.comments_sentiment"
        # API에서 가져온 ISO TIMESTAMP 문자열을 BigQuery로 입력
        errors = bq_client.insert_rows_json(comments_table_id, all_comments)

        if errors == []:
            print(f"✅ 'comments_sentiment' 테이블에 총 {len(all_comments)}개 댓글 정보 적재 완료")
        else:
            print(f"❌ 댓글 적재 중 오류 발생: {errors}")
    else:
        print("⚠️ 수집된 댓글이 없습니다.")

print("\n🎉 모든 데이터 수집 및 BigQuery 입력 작업이 완료되었습니다!")
```

```python
# 1. 필요 라이브러리 설치
!pip install --upgrade google-cloud-language google-cloud-bigquery

from datetime import datetime, timezone
from google.colab import auth
from google.cloud import bigquery, language_v1

# 구글 계정 인증
auth.authenticate_user()

# 2. 기본 설정
PROJECT_ID = "project-bb3dade7-528c-44c4-936"
DATASET_ID = "youtube_analysis"
TABLE_ID = f"{PROJECT_ID}.{DATASET_ID}.comments_sentiment"

# 클라이언트 생성
bq_client = bigquery.Client(project=PROJECT_ID)
nlp_client = language_v1.LanguageServiceClient()


# 3. Cloud Natural Language API 감성 분석 함수
def analyze_comment_sentiment(text):
    """텍스트의 감성을 분석하여 score, magnitude, label 반환"""
    if not text or not text.strip():
        return 0.0, 0.0, "NEUTRAL"

    document = language_v1.Document(
        content=text, type_=language_v1.Document.Type.PLAIN_TEXT
    )

    try:
        sentiment = nlp_client.analyze_sentiment(
            request={"document": document}
        ).document_sentiment
        score = sentiment.score  # -1.0 (부정) ~ 1.0 (긍정)
        magnitude = sentiment.magnitude  # 감정의 강도 (0.0 ~ )

        # score 기준 감성 라벨 지정
        if score >= 0.25:
            label = "POSITIVE"
        elif score <= -0.25:
            label = "NEGATIVE"
        else:
            label = "NEUTRAL"

        return score, magnitude, label
    except Exception as e:
        print(f"⚠️ 감성 분석 에러: {e}")
        return 0.0, 0.0, "NEUTRAL"


# 4. BigQuery에서 미분석 댓글 조회
print("🔍 분석할 댓글 데이터를 BigQuery에서 조회 중...")
select_query = f"""
    SELECT comment_id, text_original
    FROM `{TABLE_ID}`
    WHERE sentiment_label IS NULL
    LIMIT 500  -- API 할당량 조절을 위해 한 번에 처리할 개수
"""

query_job = bq_client.query(select_query)
rows = list(query_job.result())

if not rows:
    print("✅ 이미 모든 댓글의 감성 분석이 완료되었거나 분석할 댓글이 없습니다.")
else:
    print(f"💬 총 {len(rows)}개의 댓글에 대해 감성 분석을 시작합니다...\n")

    # 5. 댓글별 감성 분석 및 BigQuery 업데이트
    success_count = 0
    now_iso = datetime.now(timezone.utc).isoformat()

    for idx, row in enumerate(rows, 1):
        comment_id = row["comment_id"]
        text = row["text_original"]

        # 감성 분석 수행
        score, magnitude, label = analyze_comment_sentiment(text)

        # BigQuery UPDATE 쿼리 실행 (특수문자 이스케이프 처리)
        update_query = f"""
            UPDATE `{TABLE_ID}`
            SET
                sentiment_label = @label,
                sentiment_score = @score,
                sentiment_magnitude = @magnitude,
                analyzed_at = TIMESTAMP(@analyzed_at)
            WHERE comment_id = @comment_id
        """

        job_config = bigquery.QueryJobConfig(
            query_parameters=[
                bigquery.ScalarQueryParameter("label", "STRING", label),
                bigquery.ScalarQueryParameter("score", "FLOAT64", score),
                bigquery.ScalarQueryParameter("magnitude", "FLOAT64", magnitude),
                bigquery.ScalarQueryParameter("analyzed_at", "STRING", now_iso),
                bigquery.ScalarQueryParameter("comment_id", "STRING", comment_id),
            ]
        )

        bq_client.query(update_query, job_config=job_config).result()
        success_count += 1

        if idx % 10 == 0 or idx == len(rows):
            print(f" progress: [{idx}/{len(rows)}]개 댓글 처리 완료")

    print(
        f"\n🎉 성공적으로 {success_count}개 댓글의 감성 분석 결과를 BigQuery에 업데이트했습니다!"
    )
```

```python
# 1. 필요 라이브러리 설치 및 Import
!pip install --upgrade google-cloud-language google-cloud-bigquery matplotlib seaborn

from datetime import datetime, timezone
import matplotlib.pyplot as plt
from google.colab import auth
from google.cloud import bigquery, language_v1
import pandas as pd
import seaborn as sns

# 구글 계정 인증
auth.authenticate_user()

# 2. 기본 설정
PROJECT_ID = "project-bb3dade7-528c-44c4-936"
DATASET_ID = "youtube_analysis"
SOURCE_TABLE = f"{PROJECT_ID}.{DATASET_ID}.comments_sentiment"
TARGET_TABLE = f"{PROJECT_ID}.{DATASET_ID}.comments_sentiment_results"

bq_client = bigquery.Client(project=PROJECT_ID)
nlp_client = language_v1.LanguageServiceClient()


# 3. Cloud Natural Language API 감성 분석 함수
def analyze_comment_sentiment(text):
    if not text or not text.strip():
        return 0.0, 0.0, "NEUTRAL"

    document = language_v1.Document(
        content=text, type_=language_v1.Document.Type.PLAIN_TEXT
    )

    try:
        sentiment = nlp_client.analyze_sentiment(
            request={"document": document}
        ).document_sentiment
        score = sentiment.score  # -1.0(부정) ~ 1.0(긍정)
        magnitude = sentiment.magnitude  # 감정의 강도

        if score >= 0.25:
            label = "POSITIVE"
        elif score <= -0.25:
            label = "NEGATIVE"
        else:
            label = "NEUTRAL"

        return score, magnitude, label
    except Exception as e:
        return 0.0, 0.0, "NEUTRAL"


# 4. BigQuery에서 데이터 조회 후 감성 분석 수행
print("🔍 BigQuery 원본 댓글 조회 중...")
query = f"SELECT comment_id, video_id, text_original, author_name, published_at, like_count FROM `{SOURCE_TABLE}`"
rows = list(bq_client.query(query).result())

analyzed_rows = []
now_iso = datetime.now(timezone.utc).isoformat()

print(f"💬 총 {len(rows)}개 댓글 감성 분석 진행 중...")
for row in rows:
    score, magnitude, label = analyze_comment_sentiment(row["text_original"])
    analyzed_rows.append(
        {
            "comment_id": row["comment_id"],
            "video_id": row["video_id"],
            "text_original": row["text_original"],
            "author_name": row["author_name"],
            "published_at": row["published_at"].isoformat()
            if row["published_at"]
            else None,
            "like_count": row["like_count"],
            "sentiment_label": label,
            "sentiment_score": score,
            "sentiment_magnitude": magnitude,
            "analyzed_at": now_iso,
        }
    )

# 5. BigQuery에 분석 결과 업데이트(저장)
job_config = bigquery.LoadJobConfig(
    write_disposition=bigquery.WriteDisposition.WRITE_TRUNCATE,
    autodetect=True,
)
load_job = bq_client.load_table_from_json(
    analyzed_rows, TARGET_TABLE, job_config=job_config
)
load_job.result()
print(f"✅ 분석 결과가 BigQuery 테이블(`{TARGET_TABLE}`)에 업데이트되었습니다.")
```

```python
# 1. 필요 라이브러리 설치 및 한글 폰트 설정
!pip install --upgrade google-cloud-bigquery textblob > /dev/null 2>&1
!sudo apt-get install -y fonts-nanum > /dev/null 2>&1
!sudo fc-cache -fv > /dev/null 2>&1

from datetime import datetime, timezone
import re
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import pandas as pd
import seaborn as sns
from google.colab import auth
from google.cloud import bigquery
from textblob import TextBlob

# 한글 폰트 적용 (그래프 깨짐 및 마이너스 기호 방지)
font_path = '/usr/share/fonts/truetype/nanum/NanumBarunGothic.ttf'
fm.fontManager.addfont(font_path)
plt.rc('font', family='NanumBarunGothic')
plt.rcParams['axes.unicode_minus'] = False

# 2. 구글 계정 인증 및 빅쿼리 클라이언트 설정
auth.authenticate_user()

PROJECT_ID = "project-bb3dade7-528c-44c4-936"
DATASET_ID = "youtube_analysis"
SOURCE_TABLE = f"{PROJECT_ID}.{DATASET_ID}.comments_sentiment"
TARGET_TABLE = f"{PROJECT_ID}.{DATASET_ID}.comments_sentiment_results"

bq_client = bigquery.Client(project=PROJECT_ID)

# 3. 한국어 어휘 확장 및 5단계(1점~5점) 세부 감성 분석 함수
def analyze_comment_sentiment_5scale(text):
    if not text or not text.strip():
        return 3.0, 3, "3점 (중립)"

    text_lower = text.lower()

    # 한국어/영어 감성 어휘 및 이모지 다각화 사전
    very_positive_keywords = ['최고', '대박', '레전드', '갓', '미쳤다', '완벽', '지렸다', '사랑', '강추', '짱', '❤️', '🔥', '👍', 'best', 'legend', 'awesome', 'love']
    positive_keywords = ['좋다', '좋아', '재밌', '감사', '유익', '굿', '괜찮', '추천', '기대', 'good', 'nice', 'like', 'great']
    negative_keywords = ['아쉽', '별로', '노잼', '글쎄', '어려움', '지루', '오류', '버그', 'bad', 'dislike']
    very_negative_keywords = ['최악', '망함', '쓰레기', '하지마', '사기', '개판', '환불', '비추', 'worst', 'terrible', 'hate', 'trash', '💩', '🤮']

    v_pos_cnt = sum(1 for kw in very_positive_keywords if kw in text_lower)
    pos_cnt = sum(1 for kw in positive_keywords if kw in text_lower)
    neg_cnt = sum(1 for kw in negative_keywords if kw in text_lower)
    v_neg_cnt = sum(1 for kw in very_negative_keywords if kw in text_lower)

    # 기본 연속형 스코어 계산 (-1.0 ~ 1.0)
    blob = TextBlob(text)
    base_score = blob.sentiment.polarity

    # 키워드 가중치 반영
    score_offset = (v_pos_cnt * 0.4 + pos_cnt * 0.2) - (v_neg_cnt * 0.4 + neg_cnt * 0.2)
    final_score = max(-1.0, min(1.0, base_score + score_offset))

    # 1점 ~ 5점 정수 스케일 변환
    if final_score >= 0.5:
        score_5scale = 5
        label = "5점 (매우 긍정)"
    elif final_score >= 0.1:
        score_5scale = 4
        label = "4점 (긍정)"
    elif final_score <= -0.5:
        score_5scale = 1
        label = "1점 (매우 부정)"
    elif final_score <= -0.1:
        score_5scale = 2
        label = "2점 (부정)"
    else:
        score_5scale = 3
        label = "3점 (중립)"

    return round(final_score, 2), score_5scale, label

# 4. BigQuery에서 데이터 조회 및 분석 진행
print("🔍 BigQuery에서 댓글 수집 데이터 가져오는 중...")
query = f"SELECT comment_id, video_id, text_original, author_name, published_at, like_count FROM `{SOURCE_TABLE}`"
rows = list(bq_client.query(query).result())

analyzed_rows = []
now_iso = datetime.now(timezone.utc).isoformat()

print(f"💬 총 {len(rows)}개 댓글에 대해 1~5점 감성 세부 분석 진행 중...\n")

for idx, row in enumerate(rows, 1):
    raw_score, score_5scale, label_5scale = analyze_comment_sentiment_5scale(row["text_original"])

    analyzed_rows.append({
        "comment_id": row["comment_id"],
        "video_id": row["video_id"],
        "text_original": row["text_original"],
        "author_name": row["author_name"],
        "published_at": row["published_at"].isoformat() if row["published_at"] else None,
        "like_count": row["like_count"],
        "sentiment_label": label_5scale,
        "sentiment_score": score_5scale,  # 1점~5점 정수값 지정
        "sentiment_magnitude": raw_score,  # 연속형 스코어 저장
        "analyzed_at": now_iso
    })

    if idx % 200 == 0 or idx == len(rows):
        print(f" 진행률: [{idx}/{len(rows)}]")

# 5. 분석 결과를 BigQuery에 업데이트 (덮어쓰기)
job_config = bigquery.LoadJobConfig(
    write_disposition=bigquery.WriteDisposition.WRITE_TRUNCATE,
    autodetect=True,
)

print("\n⏳ BigQuery 결과 테이블(`comments_sentiment_results`) 업데이트 중...")
load_job = bq_client.load_table_from_json(analyzed_rows, TARGET_TABLE, job_config=job_config)
load_job.result()
print("✅ BigQuery 저장 완료!")

# 6. 세부 5단계 시각화 그래프 출력
df_results = pd.DataFrame(analyzed_rows)

print("\n" + "="*50)
print("📊 [오디세이 댓글 감성 점수(1점~5점) 분포 요약]")
print("="*50)
summary = df_results['sentiment_score'].value_counts().sort_index(ascending=False)
for score, count in summary.items():
    labels = {5: "5점 (매우 긍정)", 4: "4점 (긍정)", 3: "3점 (중립)", 2: "2점 (부정)", 1: "1점 (매우 부정)"}
    print(f" • {labels.get(score, score)}: {count}개 ({count/len(df_results)*100:.1f}%)")

# 그래프 그리기 (1행 2열)
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# [왼쪽] 1점~5점 단계별 댓글 수 막대 그래프
palette_colors = {'1': '#D32F2F', '2': '#F44336', '3': '#FFC107', '4': '#8BC34A', '5': '#388E3C'}
sns.countplot(
    data=df_results,
    x='sentiment_score',
    hue='sentiment_score',
    ax=axes[0],
    palette=palette_colors,
    order=[1, 2, 3, 4, 5],
    legend=False
)
axes[0].set_title('오디세이 댓글 감성 점수 세부 분포 (1점~5점)', fontsize=14, pad=12)
axes[0].set_xlabel('감성 점수 (1점:매우부정 ~ 5점:매우긍정)', fontsize=11)
axes[0].set_ylabel('댓글 개수', fontsize=11)
axes[0].set_xticklabels(['1점\n(매우부정)', '2점\n(부정)', '3점\n(중립)', '4점\n(긍정)', '5점\n(매우긍정)'])

# 막대 상단 수치 및 비율 레이블 표기
for p in axes[0].patches:
    height = p.get_height()
    if pd.notna(height) and height > 0:
        ratio = (height / len(df_results)) * 100
        axes[0].annotate(f'{int(height)}개\n({ratio:.1f}%)',
                         (p.get_x() + p.get_width() / 2., height),
                         ha='center', va='bottom', fontsize=9, xytext=(0, 3),
                         textcoords='offset points')

# [오른쪽] 감성 점수 비율 파이 차트
score_counts = df_results['sentiment_score'].value_counts().sort_index()
pie_labels = [f'{s}점' for s in score_counts.index]
pie_colors = [palette_colors.get(str(s), '#9E9E9E') for s in score_counts.index]

axes[1].pie(
    score_counts,
    labels=pie_labels,
    autopct='%1.1f%%',
    startangle=140,
    colors=pie_colors,
    explode=[0.03] * len(score_counts),
    textprops={'fontsize': 10}
)
axes[1].set_title('감성 점수 비율 구성 (1점 ~ 5점)', fontsize=14, pad=12)

plt.tight_layout()
plt.show()
```
